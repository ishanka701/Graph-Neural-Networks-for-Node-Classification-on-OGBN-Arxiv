```python
import torch

# From a list
t1 = torch.tensor([[1, 2, 3], [4, 5, 6]])
print(t1)

# Pre-filled tensors
zeros = torch.zeros(3, 3)
ones = torch.ones(2, 4)
rand = torch.rand(2, 2)          # uniform random [0,1)
identity = torch.eye(3)          # identity matrix
print(zeros, ones, rand, identity)
```

```python
t = torch.arange(12).reshape(3, 4)
print(t)

print(t[0])          # first row
print(t[:, 1])       # second column
print(t[1, 2])       # single element
print(t[0:2, 1:3])   # sub-matrix slice
```

```python
t = torch.arange(12)
print(t.view(3, 4))       # reshape into 3x4
print(t.reshape(4, 3))    # reshape into 4x3
print(t.view(2, -1))      # -1 lets PyTorch infer the size
```

```python
a = torch.rand(2, 3)
b = torch.rand(3, 4)

c1 = torch.matmul(a, b)
c2 = a @ b               # same thing, shorthand operator
print(c1.shape)          # should be (2, 4)

# Element-wise multiplication (NOT matrix mult) for contrast
d = torch.rand(2, 3)
e = torch.rand(2, 3)
elementwise = d * e
print(elementwise.shape) # (2, 3)
```

```python
a = torch.tensor([[1, 2, 3], [4, 5, 6]])   # shape (2,3)
b = torch.tensor([10, 20, 30])              # shape (3,)

result = a + b   # b is "broadcast" across each row
print(result)
```

```python
t = torch.rand(3, 4)

print(t.sum())            # sum of all elements
print(t.sum(dim=0))       # sum down each column
print(t.sum(dim=1))       # sum across each row
print(t.mean())
print(t.max())
print(t.max(dim=1))       # returns both max values AND their indices
```

```python
if torch.cuda.is_available():
    t_cpu = torch.rand(1000, 1000)
    t_gpu = t_cpu.to('cuda')
    print("On GPU:", t_gpu.device)

    # Compare speed
    import time
    start = time.time()
    _ = t_cpu @ t_cpu
    print("CPU time:", time.time() - start)

    start = time.time()
    _ = t_gpu @ t_gpu
    torch.cuda.synchronize()
    print("GPU time:", time.time() - start)
else:
    print("No GPU available in this runtime")
```
